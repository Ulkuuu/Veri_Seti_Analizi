```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df=pd.read_csv("e_ticaret_veri_seti.csv")
```

```python
print("--- İlk 10 Satır ---")
display(df.head(10))

print("\n--- Son 10 Satır ---")
display(df.tail(10))
```

```python
print(f"Satır Sayısı: {df.shape[0]}")
print(f"Sütun Sayısı: {df.shape[1]}")
print(f"Tüm Sütun İsimleri: {df.columns.tolist()}")
```

```python
print("--- Veri Seti Bilgisi (info) ---")
df.info()

print("\n--- Temel İstatistikler (describe) ---")
display(df.describe())

sayisal_sutunlar = df.select_dtypes(include=['number']).columns.tolist()
kategorik_sutunlar = df.select_dtypes(exclude=['number']).columns.tolist()

print(f"\nSayısal Sütunlar: {sayisal_sutunlar}")
print(f"Kategorik Sütunlar: {kategorik_sutunlar}")
```

```python
# Her sütundaki eksik değer sayısı
eksik_sayilari = df.isnull().sum()
print("Sütunlardaki Eksik Değer Sayıları:\n", eksik_sayilari)

# Eksik veri oranları (%)
eksik_oranlari = (df.isnull().sum() / len(df)) * 100

# En yüksek orana sahip 5 sütun
print("\nEksik Veri Oranı En Yüksek 5 Sütun (%):\n", eksik_oranlari.sort_values(ascending=False).head(5))
```

```python
# axis=1 ile satır bazında en az bir True (eksik) olanları filtreliyoruz
eksik_satirlar = df[df.isnull().any(axis=1)]
display(eksik_satirlar.head(20))
```

```python
ilgili_sutunlar = ['indirim_orani', 'musteri_puani']

print("Doldurmadan Önce Eksik Sayıları:\n", df[ilgili_sutunlar].isnull().sum())

# Medyan ile doldurma
df['indirim_orani'] = df['indirim_orani'].fillna(df['indirim_orani'].median())
df['musteri_puani'] = df['musteri_puani'].fillna(df['musteri_puani'].median())

print("\nDoldurduktan Sonra Eksik Sayıları:\n", df[ilgili_sutunlar].isnull().sum())
```

```python
kategorik_eksikler = ['odeme_turu', 'musteri_tipi']

# .mode() bir Series döndürdüğü için .iloc[0] ile ilk değeri (en çok tekrar edeni) alıyoruz
df['odeme_turu'] = df['odeme_turu'].fillna(df['odeme_turu'].mode().iloc[0])
df['musteri_tipi'] = df['musteri_tipi'].fillna(df['musteri_tipi'].mode().iloc[0])

print("Mod ile doldurma sonrası eksik kalan var mı?\n", df[kategorik_eksikler].isnull().sum())
```

```python
df['siparis_tarihi'] = pd.to_datetime(df['siparis_tarihi'])

df['siparis_yili'] = df['siparis_tarihi'].dt.year
df['siparis_ayi'] = df['siparis_tarihi'].dt.month
df['siparis_gunu'] = df['siparis_tarihi'].dt.day
df['haftanin_gunu'] = df['siparis_tarihi'].dt.day_name() # İngilizce gün isimleri verir (Monday, vb.)

display(df[['siparis_tarihi', 'siparis_yili', 'siparis_ayi', 'siparis_gunu', 'haftanin_gunu']].head(10))
```

```python
duplicate_sayisi = df.duplicated().sum()
print(f"Temizlik Öncesi Satır Sayısı: {len(df)}")

print(f"Bulunan Duplicate Kayıt Sayısı: {duplicate_sayisi}")

# Duplicate kayıtları veri setinden kalıcı olarak çıkarma
df = df.drop_duplicates()

print(f"Temizlik Sonrası Satır Sayısı: {len(df)}")
```

```python
print(f"Temizlik Öncesi Benzersiz Şehir Sayısı: {df['sehir'].nunique()}")

# Önce baştaki/sondaki boşlukları silip küçük harfe çevirelim, sonra 'istanbul' varyasyonlarını standartlaştıralım
df['sehir'] = df['sehir'].str.strip().str.lower()
df['sehir'] = df['sehir'].replace({'istanbul': 'İstanbul', 'ıstanbul': 'İstanbul'})
df['sehir'] = df['sehir'].str.title() # İlk harfleri büyüt

print(f"Temizlik Sonrası Benzersiz Şehir Sayısı: {df['sehir'].nunique()}")
# Benzersiz değerlerin listesini görmek istersen: print(df['sehir'].unique())
```

```python
# '&' işaretini boşlukla değiştirip, birden fazla boşluk varsa tek boşluğa indirgiyoruz (regex ile)
df['kategori'] = df['kategori'].str.lower().str.replace('&', ' ').str.replace('\s+', ' ', regex=True).str.title()

print("Temizlik Sonrası Kategori Dağılımı:")

print(df['kategori'].value_counts())
```

```python
hatali_sart = (
    (df['birim_fiyat'] <= 0) |
    (df['toplam_tutar'] <= 0) |
    (df['teslimat_gunu'] < 0) |
    (df['musteri_puani'] > 5)
)

hatali_kayit_sayisi = df[hatali_sart].shape[0]
print(f"Mantıksal Hatalı Kayıt Sayısı: {hatali_kayit_sayisi}")

# Hatalı kayıtları veri setinden çıkarma (~ işareti "bu şarta uymayanları (yani doğruları) al" demektir)
df = df[~hatali_sart]
```

```python
Q1 = df['birim_fiyat'].quantile(0.25)
Q3 = df['birim_fiyat'].quantile(0.75)
IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Alt Sınır: {alt_sinir}, Üst Sınır: {ust_sinir}")

aykiri_degerler = df[(df['birim_fiyat'] < alt_sinir) | (df['birim_fiyat'] > ust_sinir)]
print(f"Aykırı Değer İçeren Kayıt Sayısı: {aykiri_degerler.shape[0]}")
```

```python
plt.figure(figsize=(12, 5))

# Histogram
plt.subplot(1, 2, 1)
sns.histplot(df['toplam_tutar'], bins=30, kde=True, color='skyblue')
plt.title('Toplam Tutar Histogramı')

# Box Plot
plt.subplot(1, 2, 2)
sns.boxplot(x=df['toplam_tutar'], color='lightgreen')
plt.title('Toplam Tutar Box Plot')

plt.tight_layout()
plt.show()
```

```python
print("--- Kategori Sütunu Value Counts ---")
print(df['kategori'].value_counts())

print("\n--- Müşteri Tipi Yüzde Dağılımı ---")
print(df['musteri_tipi'].value_counts(normalize=True) * 100)

print("\n--- Temel İstatistikler ---")
display(df[['birim_fiyat', 'toplam_tutar', 'musteri_puani']].describe())

# Count Plot çizimi
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='odeme_turu', palette='viridis')
plt.title('Ödeme Türü Dağılımı')
plt.show()

# İki spesifik sorunun cevabı
en_yuksek_ciro_kategori = df.groupby('kategori')['toplam_tutar'].mean().idxmax()
en_yuksek_puanli_tip = df.groupby('musteri_tipi')['musteri_puani'].mean().idxmax()

print(f"En yüksek ortalama toplam_tutar'a sahip kategori: {en_yuksek_ciro_kategori}")
print(f"En yüksek ortalama musteri_puani'na sahip müşteri tipi: {en_yuksek_puanli_tip}")
```
